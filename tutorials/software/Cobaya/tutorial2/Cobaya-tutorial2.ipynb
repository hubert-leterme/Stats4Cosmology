```python
import os
import urllib.request
import tarfile
# loadMCSamples is expected to be available from a previous cell (wvqLWjRCZr3j).

# Define the URL for the raw tar.gz file
url = "https://raw.githubusercontent.com/hubert-leterme/Stats4Cosmology/main/tutorials/software/Cobaya/tutorial2/LCDM_chains.tar.gz"
local_tar_path = "LCDM_chains.tar.gz"
extract_dir_name = "LCDM_chains_extracted"

# --- Download the file ---
print(f"Downloading chains from {url}...")
try:
    urllib.request.urlretrieve(url, local_tar_path)
    print(f"Chains downloaded to {local_tar_path}")
except Exception as e:
    print(f"Error downloading chains: {e}")
    print("Please check the URL or your network connection.")
    gd_sample = None # Ensure gd_sample is defined, even if empty

# --- Extract the tar.gz file ---
if 'gd_sample' not in locals() or gd_sample is not None: # Only proceed if download was successful
    os.makedirs(extract_dir_name, exist_ok=True)
    print(f"Extracting chains to {extract_dir_name}...")
    try:
        with tarfile.open(local_tar_path, "r:gz") as tar:
            tar.extractall(path=extract_dir_name)
        print("Chains extracted successfully.")
    except tarfile.ReadError as e:
        print(f"Error reading tar file: {e}. The downloaded file might be corrupted or not a valid tar.gz.")
        if os.path.exists(local_tar_path): os.remove(local_tar_path) # Clean up potentially broken file
        gd_sample = None
    except Exception as e:
        print(f"An unexpected error occurred during extraction: {e}")
        gd_sample = None

# --- Locate and Load the chains using getdist.loadMCSamples ---
if 'gd_sample' not in locals() or gd_sample is not None: # Only proceed if extraction was successful
    chain_path = None
    chain_root = None

    extracted_contents = os.listdir(extract_dir_name)

    # Try to find a subdirectory (common for getdist chain archives)
    for item in extracted_contents:
        full_item_path = os.path.join(extract_dir_name, item)
        if os.path.isdir(full_item_path):
            chain_path = full_item_path
            for f_name in os.listdir(chain_path):
                if f_name.endswith(".txt") or f_name.endswith(".chains"):
                    chain_root = f_name.split('.')[0]
                    break
            if chain_root:
                break

    # Fallback: if no subdirectory, assume chains are directly in extract_dir_name
    if not chain_root:
        chain_path = extract_dir_name
        for f_name in extracted_contents:
            if f_name.endswith(".txt") or f_name.endswith(".chains"):
                chain_root = f_name.split('.')[0]
                break

    if chain_path and chain_root:
        full_file_root = os.path.join(chain_path, chain_root)
        print(f"Attempting to load chains from '{full_file_root}'...")
        try:
            # Load the Monte Carlo samples into the 'gd_sample' variable
            gd_sample = loadMCSamples(file_root=full_file_root)
            print("Chains loaded successfully into 'gd_sample'.")
            print(f"Loaded {gd_sample.samples.shape[0]} samples with {gd_sample.paramNames.numParams()} parameters.")
        except Exception as e:
            print(f"Error loading chains using getdist.loadMCSamples: {e}")
            print("Please verify the structure of the extracted files.")
            print(f"Contents of {chain_path}: {os.listdir(chain_path)}")
            gd_sample = None
    else:
        print(f"Could not find chain files within '{extract_dir_name}'.")
        print(f"Contents of '{extract_dir_name}': {os.listdir(extract_dir_name)}")
        gd_sample = None

# Clean up: remove the downloaded tar.gz file
if os.path.exists(local_tar_path):
    os.remove(local_tar_path)
    print(f"Cleaned up temporary file: {local_tar_path}")
```
